```python
# Step 47: Standardize column names

import re

df.columns = [
    re.sub(
        r"_+",
        "_",
        re.sub(
            r"[^a-zA-Z0-9]+",
            "_",
            str(column).strip().lower()
        )
    ).strip("_")
    for column in df.columns
]

print("Standardized columns:")
print(df.columns.tolist())
```

```python
# Step 48: Identify important columns

def find_column(candidates, columns):
    for candidate in candidates:
        if candidate in columns:
            return candidate
    return None


columns = set(df.columns)

column_map = {
    "name": find_column(
        ["restaurant_name", "name", "restaurant"],
        columns
    ),
    "cuisine": find_column(
        ["cuisines", "cuisine", "cuisine_type"],
        columns
    ),
    "rating": find_column(
        ["aggregate_rating", "rating", "rate"],
        columns
    ),
    "cost": find_column(
        [
            "average_cost_for_two",
            "average_cost",
            "cost_for_two"
        ],
        columns
    ),
    "city": find_column(
        ["city", "location", "city_name"],
        columns
    )
}

print("Detected column mapping:")

for feature, column in column_map.items():
    print(f"{feature}: {column}")
```

```python
# Step 49: Verify required columns

required_features = ["name", "cuisine", "rating", "cost"]

missing_features = [
    feature
    for feature in required_features
    if column_map[feature] is None
]

if missing_features:
    print("The following features need manual mapping:")
    print(missing_features)
    print("\nActual columns:", df.columns.tolist())
else:
    print("All essential columns were identified!")
```

```python
# Step 50: Create standardized recommendation data

if missing_features:
    raise ValueError(
        "Please resolve the missing feature mappings first."
    )

recommendation_df = pd.DataFrame({
    "restaurant_name": (
        df[column_map["name"]]
        .astype("string")
        .str.strip()
    ),
    "cuisines": (
        df[column_map["cuisine"]]
        .fillna("")
        .astype(str)
        .str.strip()
    ),
    "aggregate_rating": pd.to_numeric(
        df[column_map["rating"]]
        .replace("", np.nan),
        errors="coerce"
    ),
    "average_cost_for_two": pd.to_numeric(
        df[column_map["cost"]]
        .astype(str)
        .str.replace(",", "", regex=False)
        .str.strip()
        .replace("", np.nan),
        errors="coerce"
    )
})

if column_map["city"] is not None:
    recommendation_df["city"] = (
        df[column_map["city"]]
        .astype("string")
        .str.strip()
    )

recommendation_df = recommendation_df[
    recommendation_df["restaurant_name"].notna()
    & recommendation_df["restaurant_name"].ne("")
].copy()

recommendation_df.reset_index(drop=True, inplace=True)

print("Standardized dataset created!")
print("Shape:", recommendation_df.shape)

display(recommendation_df.head())
```

```python
# Step 51: Clean cuisine descriptions

recommendation_df["cuisines"] = (
    recommendation_df["cuisines"]
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

print("Cuisine descriptions cleaned.")

print(
    "Restaurants with missing cuisine information:",
    recommendation_df["cuisines"].eq("").sum()
)
```

```python
# Step 52: Convert cuisine descriptions into numerical features

from sklearn.feature_extraction.text import TfidfVectorizer

cuisine_vectorizer = TfidfVectorizer(
    tokenizer=lambda text: [
        item.strip().lower()
        for item in text.split(",")
        if item.strip()
    ],
    token_pattern=None,
    lowercase=False
)

cuisine_matrix = cuisine_vectorizer.fit_transform(
    recommendation_df["cuisines"]
)

print("TF-IDF features created.")
print("Matrix shape:", cuisine_matrix.shape)
print(
    "Unique cuisine features:",
    len(cuisine_vectorizer.get_feature_names_out())
)
```

```python
# Step 53: Create the similarity function

from sklearn.metrics.pairwise import cosine_similarity


def calculate_cuisine_similarity(preferred_cuisine):
    preferred_cuisine = str(preferred_cuisine).strip()

    if not preferred_cuisine:
        raise ValueError("Please provide a cuisine preference.")

    user_vector = cuisine_vectorizer.transform(
        [preferred_cuisine]
    )

    scores = cosine_similarity(
        user_vector,
        cuisine_matrix
    ).flatten()

    return scores


print("Cuisine similarity function created.")
```

```python
# Step 54: Build the restaurant recommendation engine

def recommend_restaurants(
    cuisine,
    max_budget=None,
    min_rating=0,
    city=None,
    top_n=10
):
    if top_n < 1:
        raise ValueError("top_n must be at least 1.")

    if max_budget is not None and max_budget < 0:
        raise ValueError("Budget cannot be negative.")

    if not 0 <= min_rating <= 5:
        raise ValueError("Minimum rating must be between 0 and 5.")

    results = recommendation_df.copy()

    results["cuisine_similarity"] = (
        calculate_cuisine_similarity(cuisine)
    )

    # Require a known cuisine for content-based matching.
    results = results[
        results["cuisines"].str.strip().ne("")
    ].copy()

    # Apply budget constraint.
    if max_budget is not None:
        results = results[
            results["average_cost_for_two"].notna()
            & (
                results["average_cost_for_two"]
                <= max_budget
            )
        ]

    # Apply rating constraint.
    if min_rating > 0:
        results = results[
            results["aggregate_rating"].notna()
            & (
                results["aggregate_rating"]
                >= min_rating
            )
        ]

    # Apply location constraint.
    if city is not None:
        if "city" not in results.columns:
            raise ValueError(
                "No city column is available in the dataset."
            )

        results = results[
            results["city"]
            .fillna("")
            .str.casefold()
            == str(city).strip().casefold()
        ]

    # Rank primarily by cuisine similarity.
    results = results.sort_values(
        by=["cuisine_similarity", "aggregate_rating"],
        ascending=[False, False],
        na_position="last"
    )

    return results.head(top_n).reset_index(drop=True)


print("Recommendation engine is ready!")
```

```python
# Step 55: Test the recommendation engine

test_results = recommend_restaurants(
    cuisine="Italian",
    max_budget=1500,
    min_rating=3.5,
    top_n=10
)

display(test_results)

print("Number of recommendations:", len(test_results))
```

```python
# Step 57: Validate recommendation constraints

def validate_recommendations(
    results,
    max_budget=None,
    min_rating=0
):
    checks = {}

    checks["budget_satisfied"] = (
        True
        if max_budget is None
        else (
            results["average_cost_for_two"].notna()
            & (
                results["average_cost_for_two"]
                <= max_budget
            )
        ).all()
    )

    checks["rating_satisfied"] = (
        True
        if min_rating == 0
        else (
            results["aggregate_rating"].notna()
            & (
                results["aggregate_rating"]
                >= min_rating
            )
        ).all()
    )

    return checks


validation = validate_recommendations(
    test_results,
    max_budget=1500,
    min_rating=3.5
)

print("Recommendation validation:")
print(validation)
```

```python
# Step 64: Create recommender.py

recommender_code = r'''
from pathlib import Path

import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


PROJECT_ROOT = Path(__file__).resolve().parent.parent
DATA_PATH = PROJECT_ROOT / "data" / "Dataset .csv"


def normalize_column_name(column):
    """Normalize column names for consistent processing."""
    import re

    name = re.sub(
        r"[^a-zA-Z0-9]+",
        "_",
        str(column).strip().lower()
    )

    return name.strip("_")


def find_column(columns, candidates):
    """Find a column using common alternative names."""
    for candidate in candidates:
        if candidate in columns:
            return candidate

    return None


def load_restaurants():
    """Load and standardize restaurant data."""

    if not DATA_PATH.exists():
        raise FileNotFoundError(
            f"Restaurant dataset not found: {DATA_PATH}"
        )

    df = pd.read_csv(DATA_PATH)

    df.columns = [
        normalize_column_name(column)
        for column in df.columns
    ]

    if df.columns.duplicated().any():
        raise ValueError(
            "Multiple dataset columns have identical names "
            "after normalization. Please resolve them."
        )

    columns = set(df.columns)

    column_map = {
        "name": find_column(
            columns,
            ["restaurant_name", "name", "restaurant"]
        ),
        "cuisine": find_column(
            columns,
            ["cuisines", "cuisine", "cuisine_type"]
        ),
        "rating": find_column(
            columns,
            ["aggregate_rating", "rating", "rate"]
        ),
        "cost": find_column(
            columns,
            [
                "average_cost_for_two",
                "average_cost",
                "cost_for_two"
            ]
        ),
        "city": find_column(
            columns,
            ["city", "city_name"]
        )
    }

    missing = [
        key
        for key in ["name", "cuisine", "rating", "cost"]
        if column_map[key] is None
    ]

    if missing:
        raise ValueError(
            f"Could not identify required columns: {missing}. "
            f"Available columns: {list(df.columns)}"
        )

    restaurants = pd.DataFrame({
        "restaurant_name": (
            df[column_map["name"]]
            .astype("string")
            .str.strip()
        ),
        "cuisines": (
            df[column_map["cuisine"]]
            .fillna("")
            .astype(str)
            .str.strip()
        ),
        "aggregate_rating": pd.to_numeric(
            df[column_map["rating"]]
            .replace("", pd.NA),
            errors="coerce"
        ),
        "average_cost_for_two": pd.to_numeric(
            df[column_map["cost"]]
            .astype(str)
            .str.replace(",", "", regex=False)
            .str.strip()
            .replace("", pd.NA),
            errors="coerce"
        )
    })

    if column_map["city"] is not None:
        restaurants["city"] = (
            df[column_map["city"]]
            .fillna("")
            .astype(str)
            .str.strip()
        )

    restaurants = restaurants.drop_duplicates().copy()

    restaurants = restaurants[
        restaurants["restaurant_name"].notna()
        & restaurants["restaurant_name"].ne("")
    ].copy()

    restaurants.reset_index(drop=True, inplace=True)

    # Ignore impossible values rather than recommending them.
    restaurants.loc[
        ~restaurants["aggregate_rating"].between(0, 5),
        "aggregate_rating"
    ] = float("nan")

    restaurants.loc[
        restaurants["average_cost_for_two"] < 0,
        "average_cost_for_two"
    ] = float("nan")

    if not restaurants["cuisines"].str.strip().ne("").any():
        raise ValueError(
            "No usable cuisine descriptions were found."
        )

    return restaurants


def build_cuisine_features(restaurants):
    """Build a TF-IDF representation of cuisine descriptions."""

    vectorizer = TfidfVectorizer(
        tokenizer=lambda text: [
            item.strip().lower()
            for item in text.split(",")
            if item.strip()
        ],
        token_pattern=None,
        lowercase=False
    )

    matrix = vectorizer.fit_transform(
        restaurants["cuisines"]
    )

    return vectorizer, matrix


def recommend_restaurants(
    restaurants,
    vectorizer,
    cuisine_matrix,
    cuisine,
    max_budget=None,
    min_rating=0,
    city=None,
    top_n=10
):
    """Recommend restaurants using cuisine similarity."""

    cuisine = str(cuisine).strip()

    if not cuisine:
        raise ValueError("Please enter a cuisine preference.")

    if top_n < 1:
        raise ValueError("top_n must be at least 1.")

    if max_budget is not None and max_budget < 0:
        raise ValueError("Budget cannot be negative.")

    if not 0 <= min_rating <= 5:
        raise ValueError("Rating must be between 0 and 5.")

    user_vector = vectorizer.transform([cuisine])

    scores = cosine_similarity(
        user_vector,
        cuisine_matrix
    ).flatten()

    results = restaurants.copy()
    results["cuisine_similarity"] = scores

    results = results[
        results["cuisines"].str.strip().ne("")
        & (results["cuisine_similarity"] > 0)
    ].copy()

    if max_budget is not None:
        results = results[
            results["average_cost_for_two"].notna()
            & (
                results["average_cost_for_two"]
                <= max_budget
            )
        ]

    if min_rating > 0:
        results = results[
            results["aggregate_rating"].notna()
            & (
                results["aggregate_rating"]
                >= min_rating
            )
        ]

    if city:
        if "city" not in results.columns:
            raise ValueError(
                "City filtering is unavailable in this dataset."
            )

        results = results[
            results["city"].str.casefold()
            == city.strip().casefold()
        ]

    results = results.sort_values(
        by=["cuisine_similarity", "aggregate_rating"],
        ascending=[False, False],
        na_position="last"
    )

    return results.head(top_n).reset_index(drop=True)
'''

RECOMMENDER_PATH = SRC_DIR / "recommender.py"

RECOMMENDER_PATH.write_text(
    recommender_code,
    encoding="utf-8"
)

print("Recommendation engine file created!")
print(RECOMMENDER_PATH)
```

```python
# Step 69: Create the Streamlit application

APP_PATH = PROJECT_ROOT / "app.py"

app_code = r'''
import streamlit as st

from recommender import (
    load_restaurants,
    build_cuisine_features,
    recommend_restaurants
)


st.set_page_config(
    page_title="Restaurant Recommendation System",
    page_icon="🍽️",
    layout="wide"
)


@st.cache_data
def load_data():
    return load_restaurants()


@st.cache_resource
def load_features():
    restaurants = load_data()

    vectorizer, matrix = build_cuisine_features(
        restaurants
    )

    return vectorizer, matrix


st.title("🍽️ Restaurant Recommendation System")

st.write(
    "Discover restaurants based on your preferred cuisine, "
    "budget, rating, and location."
)

try:
    restaurants = load_data()
    vectorizer, cuisine_matrix = load_features()

except Exception as error:
    st.error(
        "The restaurant data could not be loaded. "
        "Check your dataset and column names."
    )

    st.exception(error)
    st.stop()


st.sidebar.header("Your Preferences")

cuisines = sorted({
    cuisine.strip()
    for description in restaurants["cuisines"]
    if description.strip()
    for cuisine in description.split(",")
    if cuisine.strip()
})

if not cuisines:
    st.error("No cuisine options are available.")
    st.stop()


preferred_cuisine = st.sidebar.selectbox(
    "Preferred cuisine",
    cuisines
)


valid_costs = restaurants[
    "average_cost_for_two"].dropna()

if not valid_costs.empty:
    maximum_available_cost = int(
        valid_costs.max()
    )

    maximum_budget = st.sidebar.number_input(
        "Maximum cost for two",
        min_value=0,
        max_value=max(1, maximum_available_cost),
        value=min(1500, max(1, maximum_available_cost)),
        step=100
    )

    apply_budget = st.sidebar.checkbox(
        "Apply maximum budget",
        value=True
    )

else:
    maximum_budget = None
    apply_budget = False

    st.sidebar.warning(
        "Cost information is unavailable."
    )


minimum_rating = st.sidebar.slider(
    "Minimum rating",
    min_value=0.0,
    max_value=5.0,
    value=3.0,
    step=0.5
)


selected_city = None

if "city" in restaurants.columns:
    cities = sorted({
        city.strip()
        for city in restaurants["city"]
        if city.strip()
    })

    if cities:
        selected_city = st.sidebar.selectbox(
            "Preferred city",
            ["All cities"] + cities
        )

        if selected_city == "All cities":
            selected_city = None


number_of_results = st.sidebar.slider(
    "Number of recommendations",
    min_value=5,
    max_value=20,
    value=10
)


if st.button("Find Restaurants", type="primary"):

    try:
        results = recommend_restaurants(
            restaurants=restaurants,
            vectorizer=vectorizer,
            cuisine_matrix=cuisine_matrix,
            cuisine=preferred_cuisine,
            max_budget=(
                maximum_budget
                if apply_budget
                else None
            ),
            min_rating=minimum_rating,
            city=selected_city,
            top_n=number_of_results
        )

        if results.empty:
            st.warning(
                "No restaurants match all your preferences. "
                "Try another cuisine, reduce the minimum "
                "rating, or increase your budget."
            )

        else:
            st.success(
                f"Found {len(results)} restaurant recommendations."
            )

            st.dataframe(
                results,
                use_container_width=True,
                hide_index=True
            )

            csv_data = results.to_csv(
                index=False
            ).encode("utf-8")

            st.download_button(
                label="Download recommendations as CSV",
                data=csv_data,
                file_name="restaurant_recommendations.csv",
                mime="text/csv"
            )

    except Exception as error:
        st.error(
            "Unable to generate recommendations. "
            "Please check your preferences and dataset."
        )

        st.exception(error)


st.divider()

st.caption(
    "Cognifyz Technology | Task 2: Restaurant Recommendation System"
)
'''

APP_PATH.write_text(
    app_code,
    encoding="utf-8"
)

print("Streamlit application created!")
print("Location:", APP_PATH)
```
